**IMPORT LIBRARIES**

```python
import pandas as pd
import numpy as np

# Load raw dataset
df = pd.read_csv('/content/spinny_master_raw.csv')
```

**# Feature Engineering**

```python

# Car Age
CURRENT_YEAR = 2026
df["car_age"] = CURRENT_YEAR - df["make_year"]

# Price in Lakhs
df["price_lakhs"] = (df["price"] / 100000).round(2)

# Mileage in Thousands
df["mileage_k"] = (df["mileage"] / 1000).round(1)

# Price Segments
df["price_segment"] = pd.cut(
    df["price_lakhs"],
    bins=[0, 5, 10, 20, 1000],
    labels=["Budget", "Mid Range", "Premium", "Luxury"]
)

# Age Buckets
df["age_bucket"] = pd.cut(
    df["car_age"],
    bins=[-1, 2, 5, 8, 12, 100],
    labels=["0-2", "3-5", "6-8", "9-12", "13+"]
)

# Mileage Buckets
df["mileage_bucket"] = pd.cut(
    df["mileage_k"],
    bins=[0, 20, 50, 80, 120, 1000],
    labels=["0-20k", "20-50k", "50-80k", "80-120k", "120k+"]
)
```

**# Light Cleaning**

```python
for col in ["fuel_type", "transmission", "body_type", "car_category"]:
    if col in df.columns:
        df[col] = df[col].astype(str).str.lower().str.strip()

if "shortlist_count" in df.columns:
    df["shortlist_count"] = df["shortlist_count"].fillna(0)
```

# Save Final Dataset

```python
df.to_csv('/content/spinny_cleaned_final.csv', index=False)

print("Final Dataset Shape:", df.shape)
print("Saved as spinny_cleaned_final.csv")
```
